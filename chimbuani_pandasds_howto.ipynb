## Pandas Series
### How to create, access, and delete elements within Series.

## Step 1: Create a Series

**Simple Series**  
A series is a one-dimensional dataset in Pandas. 
To create a Series we first need to import the Pandas library, which enables us to work with the dataset more efficiently.  
For a simple series, we can first declare the values in the series by writing them in a list, as shown in the code below.
Once our list is created, we then create the Series by using the **pd.Series()** method available in the Pandas library. The argument passed in the method is our one-dimensional list of data, which enables us to create the Series dataset.  
The code also prints the dataset in the Series, which confirms that our Series is indeed created. The left column indicates the index of each element of the Series. The right column indicates the value of each element of the Series.

```python
# Create a Series from a list
import pandas as pd
my_list = [10, 20, 30, 40, 50]
series_from_list = pd.Series(my_list)
print(series_from_list)
```

**Series from a NumPy array**  
We can also create a Series from a NumPy array. To do this we first import the NumPy library.
Our Pandas library is already imported in the Code cell above.
We can then declare our array using the array() built-in method.
We then use the Series() method where we pass the array we have just created as an argument.
The code below also prints the Series confirming that we have indeed created successfully.

```python
# Create a Series from a NumPy array
import numpy as np
my_array = np.array([1, 2, 3, 4, 5])
series_from_array = pd.Series(my_array)
print(series_from_array)
```

**Series from a csv data file**  
For very large datasets, we can import the data contained in a csv file and create a Series directly from the file.  
In the example below, provided the file is properly located, we first use the method pd.read_csv() to create the corresponding DataFrame.  
Since series are one-dimensional dataset, we want to indicate which specific column or row of data we want to convert into a series.  
In the code below, we want to extract the column 'Age' only and convert it into a series. The code **age_series = titanic_df['Age']** performs this action. As long as there is only one column specified in the square brackets, Pandas interprets this as the creation of a single column of data, which makes it a series.  
The code then prints the Series to confirm that it has been properly created. We notice that the Series is truncated inside as it has a very large number of data (891 to be exact). However this does not prevent us from confirming that the Series is created.
 

```python
# Create a Series from titanic dataset
import pandas as pd

# Load the Titanic dataset into a DataFrame
titanic_df = pd.read_csv("titanic.csv")

# Extract a specific column as a Series
# For example, let's extract the 'Age' column as a Series
age_series = titanic_df['Age']

# Display the Series
print(age_series)
```

## Step 2 Accessing a Series

**Accessing elements through an index label.**  
Pandas allows us to access an element of the series via its index label.  
In the 4-element series created in the code below, each data is assigned a letter index from A to D.  
We can then print the element corresponding to the desired letter. This enables us to access the number 10 in the series Series.


```python
# Accessing Elements by Index Label
import pandas as pd

# Create a Series
data = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
series = pd.Series(data)

# Access an element by index label
print(series['A'])
```

**Accessing an element through an integer index  position.**  
We can also access an element of the Series using its integer position index, as shown in the code below.  
Note that it is important to recreate the series using a list or tuple, rather than the dictionary in the code cell above, so that an integer position index is automatically assigned to each element in the Series.

```python
#Accessing Elements by Position using integer-based positional indexing i.e. starts with zero
data = [10, 20, 30, 40]
series = pd.Series(data)
print(series[0])  
```

**Accessing elements through slicing**  
We can decide to only access and display a subset of values in the dataset. To do this, depending on whether the series was created from a dictionary, a list or a tuple, we will indicate in the print() method either the index label range, or the index integer range.  
When using a series created from a list or tuple, the last integer in the range is where the extract stops. This means that the data in that position is not shown in the display.
For example, in the code below, although the argument in the second print method is **series[2:4]**, the series slicing by position will only display the elements that are in positions 2 and 3. This is because the series stops displaying the range from position 4 onwards.


```python
# Slicing by index labels
data = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
series = pd.Series(data)
print(series['A':'B'])  

# Slicing by position
data = [10, 20, 30, 40]
series = pd.Series(data)
print(series[2:4]) 
```

## Step 3: Deleting elements within a Series

You can delete elements within Series in a number of ways, here are a few: -

- **Drop**: You can use the drop() method to delete elements by specifying the index label(s) of the elements you want to remove.
- **Slicing**: You can use slicing to select on the elements you want to keep.

## Example 1 - Dropping

In the example below, we use the **drop()** method, and the argument passed is the list of index labels of the elements we want to remove.

```python
import pandas as pd

# Create a Series
data = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
series = pd.Series(data)

# Delete elements with index labels 'B' and 'C'
series = series.drop(['B', 'C'])

print(series)
```

## Example 2 - Slicing

In the example below, after creating the Series from a dictionary, we indicate which element in the Series we want to keep by listing its letter index lebel in the following format: **series = series[:'A']**

```python
import pandas as pd

# Create a Series
data = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
series = pd.Series(data)

# Delete elements with index labels 'B' and 'C' and 'D' using slicing
series = series[:'A']

print(series)
```
