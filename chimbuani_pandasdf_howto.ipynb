## Pandas DataFrames
### How to create, access, and delete elements within Pandas DataFrames.
We will be using the Titanic data set which describes the survival status of individual passengers on the Titanic. 

## Step 1: Creating a DataFrame

```python
import pandas as pd  #Importing the Pandas third-party library

# Step 1: Create a DataFrame and load the CSV file into it
titanic_df = pd.read_csv("titanic.csv")
```

To create a dataframe, we use the Pandas library by first importing it, as it is the Python third-party library that enables us to work efficiently with single-dimensional or two-dimensional datasets.  
*pd* is the short form for Pandas used to express built-in functions in the Pandas library.  
We want Pandas to generate a DataFrame from the data contained in the titanic.csv file. We know that our file has a .csv format. So we use the function **pd.read_csv()** to create the dataframe and load the CSV data into it. The code results in the creation of a DataFrame called *titanic_df*.

## Step 2: Accessing a DataFrame

There are numerous ways to access elements in a DataFrame.  
For example, the Code cell below allows us to access elements in a dataframe in three different ways:  
- **head()** is the built-in function which displays the first 5 rows of the DataFrame.
- **describe()** is the build-in function that provides a statistical summary of the DataFrame, which helps understand the data without having to read every row individually.
- **print()** also allows to access data by printing them in the output. In the code below we have also asked the code to print the data types of each column in the DataFrame

```python
# Step 2: Access and explore the dataset

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:\n")
print(titanic_df.head())

# Check data types of columns
print("\nData types for each column:\n")
print(titanic_df.dtypes)

# Summarise the dataset
print("\nSummary statistics:")
print(titanic_df.describe())
```

## Step 3: Deleting elements within a DataFrame  
\
We can delete elements within DataFrames in a number of ways, here are a few:
- **Drop Rows or Columns**: We can use the drop() method to remove rows or columns from a DataFrame. This method creates a new DataFrame with the specified rows or columns removed.
- **Filtering**: We can filter rows based on specific conditions using boolean indexing.
Note: If "Survived" is 1, it typically means the passenger survived. If "Survived" is 0, it typically means the passenger did not survive, indicating that they died.

## Example 1 - Dropping columns
In the code below, we've decided to remove the column Cabin in the DataFrame.   
For this, we use the function **drop()** and we specify the argument to be passed, which in our case will be the list of columns to remove.  
*drop_df* is the new DataFrame resulting from the operation.  
The code also displays the first 5 rows of our new DataFrame to check that they do not contain the Cabin column anymore.

```python
# Let's create a new DataFrame with the specified column removed
drop_df = titanic_df.drop(columns=['Cabin'])

# Display the new DataFrame to check "cabin" column has been removed
print("Has the column 'Cabin' been removed?:\n")
print(drop_df.head())
```

## Example 2 - Filtering
In the code below, we've decided to filter out from the column 'Age' every passenger that is above 5 years old. We have also filtered out from the column 'Survived' every passenger that has perished in the tragedy.  
This creates a new dataframe called *filtered_df*.  
The code also prints the data of the new dataframe so that we can check that only the passengers meeting the two conditions are kept in the dataset.


```python
# Let's create a new DataFrame keeping only rows that meet the specified conditions.
filtered_df = titanic_df[(titanic_df['Age'] < 5.0) & (titanic_df['Survived'] == 1)]

# Display the new DataFrame to check the filtering has been applied
print("Has the filer been applied?:\n")
print(filtered_df)
```
